EJERCICIO 1 - SUMA DE IMÁGENES

```python
import cv2
import matplotlib.pyplot as plt

# Leer imágenes

img1 = cv2.imread('1.jpg', 0)
img2 = cv2.imread('2.jpg', 0)

# Igualar tamaño

img2 = cv2.resize(img2, (img1.shape[1], img1.shape[0]))

# Mostrar imágenes

plt.figure(figsize=(10,5))

plt.subplot(1,2,1)
plt.imshow(img1, cmap='gray')
plt.title('Imagen 1')
plt.axis('off')

plt.subplot(1,2,2)
plt.imshow(img2, cmap='gray')
plt.title('Imagen 2 Redimensionada')
plt.axis('off')

plt.show()

# SUMA PONDERADA

suma1 = cv2.addWeighted(img1, 0.7, img2, 0.3, 0)

suma2 = cv2.addWeighted(img1, 0.5, img2, 0.5, 0)

# PROMEDIO

promedio = (img1 + img2) / 2

# Mostrar resultados

plt.figure(figsize=(15,5))

plt.subplot(1,3,1)
plt.imshow(suma1, cmap='gray')
plt.title('Suma 70%-30%')
plt.axis('off')

plt.subplot(1,3,2)
plt.imshow(suma2, cmap='gray')
plt.title('Suma 50%-50%')
plt.axis('off')

plt.subplot(1,3,3)
plt.imshow(promedio, cmap='gray')
plt.title('Promedio')
plt.axis('off')

plt.show()
```

EJERCICIO 2 - RESTA DE IMÁGENES

```python
import cv2
import matplotlib.pyplot as plt

# Leer imágenes en escala de grises

img1 = cv2.imread('1.jpg', 0)
img2 = cv2.imread('2.jpg', 0)

# Igualar tamaño

img2 = cv2.resize(img2, (img1.shape[1], img1.shape[0]))

# Operaciones de resta

# Resta normal
resta = img1 - img2

# Diferencia absoluta
abs_diff = cv2.absdiff(img1, img2)

# Mostrar imágenes

plt.figure(figsize=(18,5))

plt.subplot(1,4,1)
plt.imshow(img1, cmap='gray')
plt.title('Imagen 1')
plt.axis('off')

plt.subplot(1,4,2)
plt.imshow(img2, cmap='gray')
plt.title('Imagen 2')
plt.axis('off')

plt.subplot(1,4,3)
plt.imshow(resta, cmap='gray')
plt.title('Resta')
plt.axis('off')

plt.subplot(1,4,4)
plt.imshow(abs_diff, cmap='gray')
plt.title('AbsDiff')
plt.axis('off')

plt.tight_layout()
plt.show()
```

EJERCICIO 3 - MULTIPLICACIÓN POR UN FACTOR

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Leer imagen

img = cv2.imread('imagen1.jpg', 0)

# Factores de multiplicación

img_oscura = img * 0.5
img_normal = img * 1.0
img_brillante = img * 1.5
img_muy_brillante = img * 2.0

# Evitar errores (limitar valores 0-255)

img_oscura = np.clip(img_oscura, 0, 255)
img_normal = np.clip(img_normal, 0, 255)
img_brillante = np.clip(img_brillante, 0, 255)
img_muy_brillante = np.clip(img_muy_brillante, 0, 255)

# Convertir a tipo imagen
img_oscura = img_oscura.astype(np.uint8)
img_normal = img_normal.astype(np.uint8)
img_brillante = img_brillante.astype(np.uint8)
img_muy_brillante = img_muy_brillante.astype(np.uint8)

# Mostrar resultados

plt.figure(figsize=(18,5))

plt.subplot(1,5,1)
plt.imshow(img, cmap='gray')
plt.title('Original')
plt.axis('off')

plt.subplot(1,5,2)
plt.imshow(img_oscura, cmap='gray')
plt.title('x0.5 (Oscura)')
plt.axis('off')

plt.subplot(1,5,3)
plt.imshow(img_normal, cmap='gray')
plt.title('x1.0')
plt.axis('off')

plt.subplot(1,5,4)
plt.imshow(img_brillante, cmap='gray')
plt.title('x1.5')
plt.axis('off')

plt.subplot(1,5,5)
plt.imshow(img_muy_brillante, cmap='gray')
plt.title('x2.0')
plt.axis('off')

plt.tight_layout()
plt.show()
```

EJERCICIO 4 - MULTIPLICACIÓN ELEMENTO POR ELEMENTO

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Leer imágenes

img1 = cv2.imread('1.jpg', 0)
img2 = cv2.imread('2.jpg', 0)

# Ajustar tamaño

img2 = cv2.resize(img2, (img1.shape[1], img1.shape[0]))

# Multiplicación elemento por elemento

resultado = img1 * img2

# Evitar desbordamiento (0-255)

resultado = np.clip(resultado, 0, 255)

# Convertir a imagen
resultado = resultado.astype(np.uint8)

# Mostrar resultados

plt.figure(figsize=(12,4))

plt.subplot(1,3,1)
plt.imshow(img1, cmap='gray')
plt.title('Imagen 1')
plt.axis('off')

plt.subplot(1,3,2)
plt.imshow(img2, cmap='gray')
plt.title('Imagen 2')
plt.axis('off')

plt.subplot(1,3,3)
plt.imshow(resultado, cmap='gray')
plt.title('Multiplicación')
plt.axis('off')

plt.tight_layout()
plt.show()
```

```python
import cv2
import matplotlib.pyplot as plt

# Cargar imagen
img = cv2.imread('BAJO.jpg', 0)

# Verificar
if img is None:
    print('Error: imagen no encontrada')
else:

    # Calcular histograma
    hist = cv2.calcHist([img], [0], None, [256], [0, 256])

    # Mostrar imagen + histograma
    plt.figure(figsize=(10,4))

    # Imagen original
    plt.subplot(1,2,1)
    plt.imshow(img, cmap='gray')
    plt.title('Imagen original')
    plt.axis('off')

    # Histograma
    plt.subplot(1,2,2)
    plt.plot(hist)
    plt.title('Histograma original')
    plt.xlabel('Intensidad')
    plt.ylabel('Frecuencia')

    plt.tight_layout()
    plt.show()
```
